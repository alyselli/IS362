```python
import pandas as pd

# Cumulative odometer readings at the end of each day
cumulative_miles = pd.Series(
    [55, 120, 185, 250, 310],
    index=["Day 1", "Day 2", "Day 3", "Day 4", "Day 5"]
)

cumulative_miles
```

```python
daily_miles = cumulative_miles.diff()

# On Day 1, the cumulative mileage equals the miles ridden that day
daily_miles.iloc[0] = cumulative_miles.iloc[0]

daily_miles
```

```python
daily_miles = daily_miles.astype(int)
daily_miles
```

```python
results = pd.DataFrame({
    "Cumulative Miles": cumulative_miles,
    "Miles Ridden That Day": daily_miles
})

results
```
