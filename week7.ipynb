```python

import pandas as pd

ratings = {
    "Movie": [
        "Barbie",
        "Oppenheimer",
        "Dune: Part Two",
        "Wicked",
        "Sinners",
        "Obsession"
    ],
    "Alyssa": [5, 4, 3, 5, 5, 4],
    "Sarah": [4, 5, 4, 5, 4, None],
    "Michael": [3, 5, 5, 3, 5, 4],
    "Jessica": [5, 3, None, 5, 4, 5],
    "Daniel": [4, 5, 5, 4, None, 3]
}

df = pd.DataFrame(ratings)
df.set_index("Movie", inplace=True)

df
```

```python

# Save movie ratings to a CSV file
df.to_csv("movie_ratings.csv")

# Load the CSV file into a new DataFrame
movie_ratings = pd.read_csv("movie_ratings.csv", index_col="Movie")

# Display the ratings
movie_ratings
```

```python

# Calculate average rating for each person
user_averages = movie_ratings.mean(axis=0)

print("Average Rating by Person:")
print(user_averages.round(2))

# Calculate average rating for each movie
movie_averages = movie_ratings.mean(axis=1)

print("\nAverage Rating by Movie:")
print(movie_averages.round(2))
```

```python

# Normalize ratings for each person (0 to 1 scale)
normalized_ratings = (movie_ratings - movie_ratings.min()) / (movie_ratings.max() - movie_ratings.min())

# Display normalized ratings
print("Normalized Movie Ratings:")
display(normalized_ratings.round(2))

# Calculate average normalized rating for each person
print("Average Normalized Rating by Person:")
print(normalized_ratings.mean(axis=0).round(2))

# Calculate average normalized rating for each movie
print("\nAverage Normalized Rating by Movie:")
print(normalized_ratings.mean(axis=1).round(2))
```

### Conclusion

Normalizing movie ratings makes it easier to compare how different people rate movies. Some people might give almost every movie a high rating, while others are more critical. Normalization puts everyone's ratings on the same scale from 0 to 1, which helps show their preferences compared to their other ratings.

One advantage is that it makes comparisons between people more consistent. However, a disadvantage is that we lose some of the original rating information. For example, someone who gives a movie a 3 might end up with a similar normalized score as someone who gave it a 5, depending on their other ratings.

Overall, I think normalized ratings are useful when comparing people's preferences, but actual ratings are better when you want to see how much someone liked a movie on the original scale.

```python

import pandas as pd

movie_ratings = pd.read_csv("movie_ratings.csv", index_col="Movie")

display(movie_ratings)


# Standardize ratings for each person using z-scores
standardized_ratings = (
    movie_ratings - movie_ratings.mean()
) / movie_ratings.std()

# Display standardized ratings
print("Standardized Movie Ratings:")
display(standardized_ratings.round(2))

# Calculate average standardized rating for each person
print("Average Standardized Rating by Person:")
print(standardized_ratings.mean(axis=0).round(2))

# Calculate average standardized rating for each movie
print("\nAverage Standardized Rating by Movie:")
print(standardized_ratings.mean(axis=1).round(2))
```
