```python
import requests
import pandas as pd
from getpass import getpass

api_key = getpass("Enter your NYT API key: ")
```

```python

# New York Times Books API
url = "https://api.nytimes.com/svc/books/v3/lists/current/hardcover-fiction.json"

# Request data from the API
response = requests.get(
    url,
    params={"api-key": api_key},
    timeout=30
)

# Check whether the request worked
print("Status Code:", response.status_code)

response.raise_for_status()

# Convert the response to JSON
data = response.json()

# Extract the books
books = data["results"]["books"]

# Convert the JSON into a pandas DataFrame
df = pd.DataFrame(books)

# Select columns
df = df[["rank", "title", "author", "publisher", "description"]]

# Display the first 10 books
display(df.head(10))
```

```python
print("Total books retrieved:", len(df))
print("DataFrame dimensions:", df.shape)
print("Column names:", df.columns.tolist())
```
