## Question 1: What is the northernmost airport in the United States?

I first sorted the airport data by latitude from highest to lowest. 
Since higher latitude values represent locations farther north, this 
shows the northernmost airports in the dataset.

```python
northern_airports = airports.sort_values("lat", ascending=False)

northern_airports[["faa", "name", "lat", "lon"]].head(5)
```

### Answer

At first, the data showed Dillant-Hopkins Airport (EEN) as the northernmost airport. However, the coordinates in the dataset are incorrect because this airport is actually located in New Hampshire. After looking at the next results and checking the location, I found that **Wiley Post-Will Rogers Memorial Airport (BRW) in Alaska** is the northernmost airport, at about 71.29° N.

## Question 2: What is the easternmost airport in the United States?

I examined the longitude values to find the easternmost airports. 
Because longitude wraps around at the International Date Line, I 
looked at airports at both extremes of the longitude data before 
determining the answer.

```python
eastern_airports = airports.sort_values("lon", ascending=False)

eastern_airports[["faa", "name", "lat", "lon"]].head(5)
```

```python
western_longitudes = airports.sort_values("lon", ascending=True)

western_longitudes[["faa", "name", "lat", "lon"]].head(5)
```

### Answer

The longitude data had some incorrect coordinates, so I had to look at the results more closely instead of just choosing the first one. Eareckson Air Station (SYA) is located on Shemya Island, Alaska, and its longitude of about 174.11° E is valid. Because Alaska crosses into the Eastern Hemisphere, **Eareckson Air Station (SYA)** is the easternmost airport in the dataset.

## Question 3: On February 12, 2013, which New York area airport had the windiest weather?

I filtered the weather data to include only observations from 
February 12, 2013. I then compared the wind speed observations for 
the three New York area airports.

```python
feb12_weather = weather[
    (weather["year"] == 2013) &
    (weather["month"] == 2) &
    (weather["day"] == 12)
]

feb12_weather.head()
```

```python
feb12_weather[["origin", "hour", "wind_speed"]].sort_values(
    "wind_speed", ascending=False
).head(15)
```

```python
feb12_weather.groupby("origin")["wind_speed"].describe()
```

```python
feb12_weather.groupby("origin")["wind_speed"].max()
```

```python
feb12_weather.groupby("origin")["wind_speed"].mean()
```

```python
feb12_weather.groupby("origin")["wind_speed"].median()
```

### Answer

When I looked at the wind speeds, Newark (EWR) had a value of 1048.36 mph. This is clearly an outlier and not a realistic wind speed, so I did not use it when comparing the airports. After accounting for the outlier, **LaGuardia Airport (LGA)** had the highest valid wind speed at about 23.02 mph. Therefore, LGA had the windiest weather on February 12, 2013.
